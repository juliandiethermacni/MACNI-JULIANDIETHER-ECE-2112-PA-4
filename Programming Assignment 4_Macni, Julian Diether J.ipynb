```python
#imports the Pandas and Matplotlib libraries
import pandas as pd
import matplotlib.pyplot as plt
```

```python
#imports the table file and displays it
df = pd.read_excel('board2.xlsx')
df
```

```python
#gets the overall grade average from each student and adds it to the table
df['Average'] = df[['Math', 'Electronics', 'GEAS', 'Communication']].mean(axis = 1)
df
```

```python
#only displays the list of students from Visayas and has a track of communication
VisComm = df.loc[(df['Hometown'] == 'Visayas') & (df['Track'] == 'Communication'), ['Name', 'Gender', 'Math', 'Electronics', 'Average']]
VisComm
```

```python
#displays the number of rows
len(VisComm)
```

```python
#only displays the list of female students from Visayas
VisFemale = df.loc[(df['Hometown'] == 'Visayas') & (df['Gender'] == 'Female'), ['Name', 'Track', 'GEAS', 'Electronics', 'Average']]
VisFemale
```

```python
#only diplays the list of female students from Visayas that have an average of at least 60
VisFemale.loc[(VisFemale['Average'] >= 60)]
```

## C. CATEGORY-AVERAGE VISUALIZATION

```python
#gets the mean average of each track
Communication = df.loc[(df['Track'] == 'Communication'), 'Average'].mean()
Instrumentation = df.loc[(df['Track'] == 'Instrumentation'), 'Average'].mean()
Microelectronics = df.loc[(df['Track'] == 'Microelectronics'), 'Average'].mean()

Track = pd.DataFrame({'Mean Average': [Communication, Instrumentation, Microelectronics]},
                      index = ['Communication', 'Instrumentation', 'Microelectronics'])
Track
```

```python
#gets the mean average of each gender
Male = df.loc[(df['Gender'] == 'Male'), 'Average'].mean()
Female = df.loc[(df['Gender'] == 'Female'), 'Average'].mean()


Gender = pd.DataFrame({'Mean Average': [Male, Female]},
                      index = ['Male', 'Female'])
Gender
```

```python
#gets the mean average of each hometown
Luzon = df.loc[(df['Hometown'] == 'Luzon'), 'Average'].mean()
Visayas = df.loc[(df['Hometown'] == 'Visayas'), 'Average'].mean()
Mindanao = df.loc[(df['Hometown'] == 'Mindanao'), 'Average'].mean()


Hometown = pd.DataFrame({'Mean Average': [Luzon, Visayas, Mindanao]},
                      index = ['Luzon', 'Visayas', 'Mindanao'])
Hometown
```

```python
#displays bar charts of the mean average by track, gender, and hometown
plt.figure(figsize=(18, 4))

plt.subplot(1, 3, 1)
plt.bar(Track.index, Track['Mean Average'])
plt.title('Mean Average by Track')
plt.ylabel('Mean Average')

plt.subplot(1, 3, 2)
plt.bar(Gender.index, Gender['Mean Average'])
plt.title('Mean Average by Gender')

plt.subplot(1, 3, 3)
plt.bar(Hometown.index, Hometown['Mean Average'])
plt.title('Mean Average by Hometown')

plt.show()
```

**Track:** The Communication track recorded the highest sample mean among the three tracks, with an average score of 67.98.

**Gender:** Male students recorded the highest sample mean between the two genders, with an average score of 67.18.

**Hometown:** Students from Luzon recorded the highest sample mean among the three regions, with an average score of 68.08.
